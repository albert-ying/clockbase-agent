### Select the intersection of GSEs

#### mouse rna

```python
# Read the aging clock file
_df = pd.read_csv("/dfs/user/hanchenw/Agent_PG/clockbase_agent/data/clock_output/mouse_rna_age.csv")
_df
```

```python
len(_df["gse_id"].unique())
```

```python
# Read the dataset level file
gse_rna = pd.read_csv("../data/dataset_level/gse_rna_mm.csv")
gse_rna
```

```python
len((gse_rna["accession"]).unique())
```

```python
len(os.listdir("../data/GEO_mm_rna/processed_metadata_preharmonize/"))
```

```python
%%time

selected_gse_ids = []
for gse_id in _df["gse_id"].unique():
    if gse_id in gse_rna["accession"].values:
        if os.path.exists(f"../data/GEO_mm_rna/processed_metadata_preharmonize/{gse_id}.csv"):
            selected_gse_ids.append(gse_id)

print(f"Total matching entries: {len(selected_gse_ids)}")
```

```python
_df
```

```python
_ddf = _df[_df["gse_id"].isin(selected_gse_ids)][[
    "gse_id", 
    "gsm_id", 
    "Hazard_Rodents_All_All_scaleddiff_EN_tAge", 
    "Hazard_Mouse_All_All_scaleddiff_EN_tAge"
]]
_ddf
```
